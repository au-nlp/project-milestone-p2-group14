```python
import pandas as pd

df = pd.read_json("data/episodeLevelDataSample_fixed.jsonl", lines=True)

# Columns that should always be arrays
array_cols = ['hostPredictedNames', 'guestPredictedNames', 'neitherPredictedNames', 'mainEpSpeakers']
for col in array_cols:
    if col in df.columns:
        df[col] = df[col].apply(lambda x: x if isinstance(x, list) else [x])

# This column is special because we have to transform is from Dict to string because of some different values in some rows
if 'hostSpeakerLabels' in df.columns:
    df['hostSpeakerLabels'] = df['hostSpeakerLabels'].apply(lambda x: str(x) if isinstance(x, dict) else (str(x) if x is not None else ""))


# Columns that should always be strings
string_cols = [
    'transcript', 'rssUrl', 'epTitle', 'epDescription', 'mp3url', 'podTitle', 'itunesAuthor',
    'itunesOwnerName', 'imageUrl', 'language', 'host', 'podDescription', 'category1', 'category2',
    'category3', 'category4', 'category5', 'category6', 'category7', 'category8', 'category9',
    'category10', 'oldestEpisodeDate', 'guestPredictedNames', 'guestSpeakerLabels', 'overlapPropTurnCount',
    'avgTurnDuration', 'overlapPropDuration', 'totalSpLabels'
]
for col in string_cols:
    if col in df.columns:
        df[col] = df[col].apply(lambda x: str(x) if x is not None and not isinstance(x, list) and not isinstance(x, dict) else "")

# Columns that should always be numeric
numeric_cols = [
    'lastUpdate', 'explicit', 'createdOn', 'episodeDateLocalized', 'durationSeconds', 'numUniqueHosts',
    'numUniqueGuests', 'numUniqueNeithers', 'numMainSpeakers'
]
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Save back to JSONL
df.to_json("data/episodeLevelDataSample_fixed.jsonl", orient="records", lines=True)
```
